```python
import pandas as pd
import numpy as np

query = """
SELECT *
FROM sales_transactions_2022_2025
"""

df = pd.read_sql(query,conn)

print("Data loaded successfully!")
print(df.shape)
```

```python
df.isnull().sum()
```

```python
df["Customer_Name"] = (
    df["Customer_Name"]
    .fillna(
        df.groupby("Customer_ID")["Customer_Name"]
        .transform("first")
    )
)
```

```python
df["Customer_Age"] = (
    df["Customer_Age"]
    .fillna(
        df.groupby("Customer_ID")["Customer_Age"]
        .transform("first")
    )
)
```

```python
df["Customer_Gender"] = (
    df["Customer_Gender"]
    .fillna(
        df.groupby("Customer_ID")["Customer_Gender"]
        .transform("first")
    )
)
```

```python
df["Product_Name"] = (
    df["Product_Name"]
    .fillna(
        df.groupby("Product_ID")["Product_Name"]
        .transform("first")
    )
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df["Order_Hour"] = (
    pd.Timestamp("2000-01-01") + df["Order_Time"]
).dt.strftime("%H")
```

```python
df["Day"] = pd.Categorical(
    df["Day"],
    categories=["Monday", "Tuesday", "Wednesday",
                "Thursday", "Friday", "Saturday", "Sunday"],
    ordered=True
)
```

```python
plt.figure(figsize=(12, 2))
heatmap = df.pivot_table(
    index="Day",
    columns="Order_Hour",
    values="Order_ID",
    aggfunc="count"
)

sns.heatmap(heatmap)
plt.show()
```

```python
df["Customer_Rating"].corr(df["Profit"])
```

```python
daily = df.groupby("Order_Date").agg(
    Order_Count=("Order_ID", "count"),
    Profit=("Profit", "sum")
).reset_index()

avg_profit = daily.groupby("Order_Count")["Profit"].mean().reset_index()

sns.lineplot(data=avg_profit, x="Order_Count", y="Profit", marker="o")
plt.xlabel("Orders per Day")
plt.ylabel("Average Daily Profit")
plt.title("Order Count vs Average Daily Profit")
plt.show()
```

```python
pie = df["Customer_Gender"].value_counts()

pie[["Male", "Female"]].plot.pie(
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.show()
```

```python


data = df.groupby("Product_Category").agg(
    Profit=("Profit", "sum"),
    Order_Count=("Order_ID", "count")
)

data.plot(kind="bar", figsize=(10,5))

plt.title("Profit & Order Count by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Value")
plt.xticks(rotation=45)
plt.show()
```
