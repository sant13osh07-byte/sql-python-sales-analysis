```python
ax = data.groupby(["region", "product_category"])["revenue"].sum().unstack().plot.bar(cmap='Pastel1')

for container in ax.containers:
     plt.bar_label(container, fmt='%.0f', padding=2, rotation=90, fontsize=7)
```

```python
plt.figure(figsize=(14, 7))

# Layer 1: Enhanced Boxen Plot for Distribution Quantiles
sns.boxenplot(
    data=data, 
    x="region", 
    y="revenue", 
    hue="product_category", 
    palette="Set2", 
    k_depth="proportion"
)

# Layer 2: Swarmplot Overlay for Individual Data Point Density
sns.stripplot(
    data=data, 
    x="region", 
    y="revenue", 
    hue="product_category", 
    dodge=True, 
    color="black", 
    alpha=0.3, 
    size=3,
    legend=False
)

plt.title("Regional Revenue Distribution across Product Categories", fontsize=14, pad=15)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', title="Category")
plt.tight_layout()
plt.show()
```
